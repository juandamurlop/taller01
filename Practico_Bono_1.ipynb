#### MENOR DE DOS PARES: Escriba una función que devuelva el menor de dos números dados *si* ambos números son pares, pero devuelva el mayor si uno o ambos números son impares
    lesser_of_two_evens(2,4) --> 2
    lesser_of_two_evens(2,5) --> 5

```python
def lesser_of_two_evens(a,b):

    if((a%2)==0) and ((b%2)==0):    #se revisa si ambos números son pares
        if(a<b):                    #se revisa cual es el menor
            return a                
        else:                       #dependiendo el caso se retorna el menor
            return b
    else:                           #si alguno de los números es impar
        if(a<b):                    #se revisa cual es el mayor
            return b
        else:                       #dependiendo el caso se retorna el mayor
            return a
```

```python
# se revisa
lesser_of_two_evens(2,4)
```

```python
# se revisa
lesser_of_two_evens(2,5)
```

#### GALLETAS DE ANIMALES: Escribir una función que toma una cadena de dos palabras y devuelve True si ambas palabras comienzan con la misma letra
    animal_crackers('Ciudad Calor') --> True
    animal_crackers('Lugar Frio') --> False

```python
def animal_crackers(text):
  verif = False                     #se inicia la variable verif en False
  N = len(text)                     #se obtiene la longitud del texto

  for i in range(N):                #ciclo para recorrer el texto
    if(text[i] == " "):             #cuando se encuentra un espacio en el texto
      if(text[0] == text[i + 1]):   #si la primera letra y la que le sigue al espacio son iguales continuamos
        verif = True                #se cambia el valor de verif a True
        break                       
  return verif                      #se retorna el valor de verif
```

```python
# se revisa
animal_crackers('Ciudad Calor')
```

```python
# se revisa
animal_crackers('Lugar Frio')
```

#### HACE VEINTE: dados dos enteros, devuelve True si la suma de los enteros es 20 o si uno de los enteros es 20. De lo contrario, devuelve False

    makes_twenty(20,10) --> True
    makes_twenty(12,8) --> True
    makes_twenty(2,3) --> False

```python
def makes_twenty(n1,n2):
  if((n1 + n2 == 20) or ((n1 == 20) or (n2 == 20))):    #se revisa si la suma de los dos números es 20 o si alguno de los dos números es 20
    return True                                         #Si se cumple la condición es True
  else:                                                 #Si la suma, o alguno de los dos números no es 20 entra al bloque else
    return False                                        #si no se cumple la condición es False
```

```python
# se revisa
makes_twenty(20,10)
```

```python
# se revisa
makes_twenty(2,3)
```

```python
makes_twenty(12,8)
# se revisa
```

#### MAYÚSCULAS: Escribir una función que haga mayúscula la primera y cuarta letra de una cadena de texto
     
    old_macdonald('macdonald') --> MacDonald
    
Note: `'macdonald'.capitalize()` returns `'Macdonald'`

```python
def old_macdonald(name):

    N = len(name)                   #se obtiene la longitud del nombre
    letra1 = name[0:3]              #se genera una variable con las tres primeras letras del nombre
    letra2 = name[3:N]              #se genera una variable con las letras restantes del nombre

    letra1 = letra1.capitalize()    #se pone en mayuscula la primera letra
    letra2 = letra2.capitalize()    #se pone en mayuscula la primera letra de la segunda parte del nombre

    resultado = letra1 + letra2     #concateno ambas partes
    return resultado                #retornamos el resultado
```

```python
# Check
old_macdonald('macdonald')
```

#### REVERSA: Dada una oración, retornar la misma oración pero en reversa

    master_yoda('I am home') --> 'home am I'
    master_yoda('We are ready') --> 'ready are We'
    
Note: Revisar método .join()

    >>> "--".join(['a','b','c'])
    >>> 'a--b--c'

Resultado de .join():

    >>> " ".join(['Hello','world'])
    >>> "Hello world"

```python
def master_yoda(text):
    
    palabras = text.split(" ")              #a una variable le asignamos el texto dividido por los espacios
    palabras.reverse()                      #a la lista resultante la ponemos en el orden inverso
    reversa = " ".join(palabras)            #usamos el metodo join para agrupar la lista con espacios en el medio
    return reversa
```

```python
# Revisión
master_yoda('We are ready')
```

#### PROBLEMA 33: 

Dada una lista de enteros, retornar True si el arreglo tinene un  3  seguido por un  3 .

    has_33([1, 3, 3]) → True
    has_33([1, 3, 1, 3]) → False
    has_33([3, 1, 3]) → False

```python
def has_33(nums):

    N = len(nums)                                   #Definimos el tamaño de la lista
    ver = False                                     #verificación para parar el bucle

    for i in range(N):                              #ciclo para recorrer la lista
        if(i + 1 == N):                             #Si llegamos al ultimo lugar de la lista detenemos el ciclo
            break
        elif(nums[i] == 3) and (nums[i+1] == 3):    #Si en la posición actual y en la siguiente hay un 3, devolvemos el True
            ver = True
            break

    return ver
```

```python
# Revisión
has_33([1, 3, 3])
```

```python
# Revisión
has_33([1, 3, 1, 3])
```

```python
# Revisión
has_33([3, 1, 3])
```

#### REPLICADOR: Dada una cadena, devuelve una cadena donde por cada carácter en el original hay tres caracteres REPETIDOS
    paper_doll('Hello') --> 'HHHeeellllllooo'
    paper_doll('Mississippi') --> 'MMMiiissssssiiippppppiii'

```python
def paper_doll(text):

    N = len(text)                               #Obtenemos la longitud del texto
    palabra = ""                                #creamos el texto resultante
    for i in range(N):                          #recorremos la palabra que entró
        palabra = palabra + (3 * text[i])       #la palabra resultante será la misma palabra concatenada con 3 veces la letra...
    return palabra                                              #en la que nos encontremos
        
```

```python
# se revisa
paper_doll('Hello')
```

```python
# se revisa
paper_doll('Mississippi')
```

#### BLACKJACK: Dados tres enteros entre 1 y 11, si su suma es menor o igual a 21, devuelve su suma. Si su suma excede 21 *y* hay un once, reduce la suma total en 10. Finalmente, si la suma (incluso después del ajuste) excede 21, devuelve 'BUST'
    blackjack(5,6,7) --> 18
    blackjack(9,9,9) --> 'BUST'
    blackjack(9,9,11) --> 19

```python
def blackjack(a,b,c): #se define la función blackjack con tres parámetros a, b y c
    sum = a + b + c  #suma de los tres números

    if(sum > 21):  #si la suma es mayor a 21 se entra al bloque if
        if(a == 11) or (b == 11) or (c == 11):
            sum = sum - 10
        
    if(sum>21):  
        return 'BUST'
    else:
        return sum
    
```

```python
# se revisa
blackjack(5,6,7)
```

```python
# se revisa
blackjack(9,9,9)
```

```python
# se revisa
blackjack(9,9,11)
```
