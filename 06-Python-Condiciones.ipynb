<h2>Introducción</h2>
En programacion, las condiciones permiten que un programa tome decisiones en función de ciertas condiciones o criterios establecidos. Estas estructuras de control facilitan la ejecución de diferentes bloques de código según se cumplan o no determinadas expresiones lógicas, haciendo que los programas puedan adaptarse a distintas situaciones.

ademas permiten controlar el flujo de ejecución según los datos y el contexto. Esta capacidad de tomar decisiones incrementa la flexibilidad y la funcionalidad de los programas.


Escribe una declaración if que determine si un disco musical tiene una calificación mayor a 8. Compruébalo utilizando el calificación del disco <b>“Back in Black”</b> el cual tiene un valor de 8.5. Si la condición es verdadera, imprime “"This album is Amazing!"”

```python
BackinBlack = 8.5

if(BackinBlack > 8 ):
    print("This album is Amazing!")
#se crea una variable llamada "BackinBlack" luego se evalua si su valor es mayor a 8, si es así se imprime el mensaje 
```

Escribe una declaración if-else que determine lo siguiente. Si la calificación es mayor que 8 imprime “this album is amazing”. Si es menor o igual a 8 que imprima “this album is ok”.

```python
rate = 7

if(rate > 8):
    print("This album is Amazing!")         #Si la calificación es mayor a 8 se imprime "Amazing"
else:
    print("This album is ok")               #en cualquier otro escenario se imprime que el album está "OK"
```
