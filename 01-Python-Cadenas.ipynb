<h2>Introducción</h2>
En programacion, una cadena es un tipo de dato importante para representar información textual por una secuencia de caracteres. se usa para almacenar, manipular y presentar datos como mensajes, nombres y entradas del usuario, permitiendo aplicar varias operaciones para el tratamiento y analisis del texto. 


<h2>Ejercicios Cadenas Python</h2>

¿Cual es el valor de la variable <code>A</code> después de ejecutar el siguiente código?

```python

A = "1"
A
#la variable A no guardo el 1 como un numero, sino como una variable de texto.
```

¿Cual es el valor de la variable <code>B</code> después de ejecutar el siguiente código?

```python
B = "2"
B
#la variable B no guardo el 2 como un numero, sino como una variable de texto
```

¿Cual es el valor de la variable <code>C</code> después de ejecutar el siguiente código?

```python
C = A + B
C

#si intentamos sumar A y B nos va a dar la concatenación de ambos números, ya que no son números sino cadenas de texto
```

En la variable <code>D</code> haz uso del slicing para imprimir solo los primeros tres elementos:

```python
D = "ABCDEFG"
print(D[0:3])
#El slicing [0:3] nos da los elementos de la cadena que tienen indice del 0 al 2
```

Usa un valor de stride de 2 para imprimir cada segundo elemento de la cadena <code>E</code>:

```python
E = "clocrkr1e1c1t"
print(E[::2])
#usando el stride con valor 2, se seleccionara cada segundo caracter de la cadena
```

Imprime una barra invertida:

```python
print("\\")
print(r"\ ")
#Para imprimir correctamente la barra invertida se puede hacer de dos formas
#1. ponerla dos veces para que se imprima correctamente
#2. poner una "r" antes de la cadena a imprimir, para que python imprima tal cual lo que aparece ignorando si hay algo despues de la barra invertida
```

Convierte el mayúsculas el contenido de la variable <code>F</code>:

```python
F = "You are wrong"
print(F.upper())
#Al usar el metodo upper se transforma a mayusculas la cadena en la que se esté usando
```

En la variable <code>G</code>, encuentra el primer índice de la sub-cadena <code>snow</code>:

```python
G = "Mary had a little lamb Little lamb, little lamb Mary had a little lamb \
Its fleece was white as snow And everywhere that Mary went Mary went, Mary went \
Everywhere that Mary went The lamb was sure to go"

print(G.find("snow"))
#Usando el metodo "find" podemos encontrar en que posición se encuentra la sub-cadena que se requiera
   #en este caso "snow" se encontró en la posición 95
```

En la variable <code>G</code>, reemplaza la sub-cadena <code>Mary</code> con <code>Bob</code>:

```python
print(G.replace('Mary', 'Bob'))
#Usando el metodo "replace" se puede reemplazar todas las sub-cadenas en las que se encuentre "Mary" por "Bob"
```

<h2>Conclusiones</h2>

En este cuaderno se practicaron distintas operaciones basicas sobre cadenas de texto en Python, lo que permitio comprender como se comportan y cómo pueden transformarse segun lo que se requiera realizar con ellas, las cadenas en Python son inmutables, por lo que las transformaciones que se les aplican no modifican la cadena original, sino que generan una nueva. Por ejemplo, si a una cadena A se le aplica el método A.replace('am', 'pm') y luego se imprime A, los valores originales permanecen sin cambios, ya que el reemplazo no afecta directamente a la cadena inicial.
