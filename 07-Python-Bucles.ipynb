<h2>Introducción</h2>
En programacion , los bucles permiten ejecutar repetidamente un bloque de instrucciones mientras se cumpla una condición determinada o se recorra una secuencia de elementos. Su uso facilita la automatización de tareas repetitivas y el procesamiento eficiente de colecciones de datos.

Escribe un bucle <code>for</code> que imprima todos los elementos entre <b>-5</b> y <b>5</b> usando la función range.

```python
for i in range(-5, 6): #el bucle recorre desde -5 hasta 5, ya que el 6 no se incluye
    print(i)
```

Imprime los elementos de la siguiente lista:
<code>Genres=[ 'rock', 'R&B', 'Soundtrack', 'R&B', 'soul', 'pop']</code>
Y asegúrate de seguir las convenciones del lenguaje Python

```python
Genres=[ 'rock', 'R&B', 'Soundtrack', 'R&B', 'soul', 'pop']     #se genero la lista de generos musicales para el desarrollo del ejercicio
N = len(Genres)                                                 #tenemos el tamaño de la lista para poder recorrerla con el ciclo for

for i in range(N):                                              #en range(N) se recorre la lista de generos musicales
    print(Genres[i])                                            #el ciclo nos da acceso a cada elemento de la lista y lo imprime en pantalla
```

Escribe un bucle for que imprima la siguiente lista: <code>squares=['red', 'yellow', 'green', 'purple', 'blue']</code>

```python
squares=['red', 'yellow', 'green', 'purple', 'blue']

for color in squares:   #se imprime cada color de la lista squares, ya que el ciclo for recorre la lista y va asignando a la variable color cada elemento de la lista
    print(color)            
```

Escribe un bucle while que muestre los valores de la calificación de las canciones de un disco almacenados en la lista <code>PlayListRatings</code>. Si la calificación es menor a 6, sal del bucle. La lista <code>PlayListRatings</code> esta compuesta por: <code>PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]</code>

```python
PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]

calification = PlayListRatings[0] #se inicia la variable calification con el primer valor de la lista PlayListRatings
i = 1   #la lista se recorre desde la posición 1 ya que la posición 0 ya fue asignada a la variable calification

while(calification >= 6):    #si es menor a 6 se sale del bucle, si es mayor o igual a 6 se sigue recorriendo la lista
    print("rating", i, "es", calification)  #se muestra el valor de la calificacion y el indice en el que se encuentra
    calification = PlayListRatings[i]   #se asigna a la variable calification el valor de la lista en la posición i
    i = i + 1   #se va sumando 1 a la variable i para recorrer la lista

print("Rating menor a 6 fue encontrado en la posición", i, "con indice", i - 1) 
        #se muestra el indice en el que se encontró la calificación menor a 6, se resta 1 ya que la variable i se incrementa antes de salir del bucle
```

Escribe un bucle while que copie las cadenas <code>'orange'</code> de la lista <code>squares</code> a la lista <code>new_squares</code>. Detente y sal del ciclo si el valor en la lista no es <code>'orange'</code>:

```python
squares = ['orange', 'orange', 'purple', 'blue ', 'orange']
new_squares = []

copy = squares[0]   #se copia el primer elemento de la lista squares a la variable copy
i = 1   #la variable inicia en 1

while(copy == 'orange'):    #si el elemento a copiar es igual a 'orange' se entra al bucle, si no es igual se sale del bucle
    new_squares.append(copy)    #a la lista nueva agragamos el elemento a copiar
    copy = squares[i]       #ahora el siguiente elemento de la lista es asignado al elemento a copiar
    i = i + 1               #agregamos uno a la variable para movernos en la lista

print(new_squares)
```

El desarrollo de este cuaderno permitió entender las diferencias de sintaxis de los ciclos en Python, las cuales varían frente a los lenguajes trabajados anteriormente y pueden resultar confusas en un primer momento. Sin embargo, en comparación con otros lenguajes, Python utiliza una sintaxis más corta e incluso más “automatizada”, ya que no es necesario controlar de forma tan manual el recorrido del ciclo, como el metodo range y de los ciclos sobre listas. En lenguajes como Java o C++, es necesario crear una variable que controle el recorrido y definir manualmente cuándo inicia y termina el ciclo. En Python, basta con indicar la lista o el rango, y el lenguaje se encarga automáticamente del inicio y del final del recorrido.
