<h2>Introducción</h2>
En Python, un diccionario es una estructura de datos que permite almacenar información en pares de clave y valor. Esta forma de organización facilita el acceso a los datos, ya que cada valor se asocia a una clave única que permite identificarlo de manera directa dentro de la estructura.


```python
soundtrack_dic = {"The Bodyguard":"1992", "Saturday Night Fever":"1977"}    #se generael diccionario
soundtrack_dic 
```

a) En el diccionario <code>soundtrack_dict</code>. ¿Cuáles son las llaves?

```python
soundtrack_dic.keys()       #las llaves del diccionario
```

b) En el diccionario <code>soundtrack_dict</code>. ¿Cuáles son los valores?

```python
soundtrack_dic.values()     #devuelve los valores del diccionario
```

<b>Necesitarás este diccionario para las siguientes preguntas:</b>

Los discos <b>Back in Black</b>, <b>The Bodyguard</b> y <b>Thriller</b> poseen el siguiente record de ventas en millones, 50, 50 y 65 respectivamente:

a) Crea un diccionario <code>album_sales_dict</code> donde las llaves sean el nombre del disco y los valores la cantidad de ventas en millones.

```python
dic_ventas = {"back in black": 50, "the bodyguard": 50, "Thriller": 65}
dic_ventas
#se genera un diccionario para almacenar las ventas de los discos "back in black", "the bodyguard" y "Thriller" con sus respectivas cantidades vendidas.
```

b) Usa el diccionario para encontrar las ventas totales de <b>Thriller</b>:

```python
print("las ventas de Thriller son de", dic_ventas["Thriller"], "millones")
#en el diccionario hacemos uso de "[]" para buscar la llave "Thriller" y sacar su información
```

c) Encuentra los nombres de los discos dentro del diccionario mediante el método <code>keys</code>:

```python
dic_ventas.keys() 
#en el diccionario usamos el metodo keys para obtener los discos
```

d) Encuentra los records de ventas dentro del diccionario mediante el método <code>values</code>:

```python
dic_ventas.values()
#usamos el metodo values para obtener las ventas de los discos
```
