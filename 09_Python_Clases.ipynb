<h2>Introducción</h2>
En programacion, los objetos son entidades que representan elementos del mundo real o conceptos abstractos dentro de un programa. Cada objeto combina información, mediante atributos, y comportamiento, mediante métodos, permitiendo modelar de manera organizada los elementos con los que se trabaja en una aplicacion.

El cuaderno enviado no cuenta con ejercicios enfocados en clases, asi que decidí crear una clase personas y empezar a experimentar con ella para practicar.

```python
class Persona(object):                  #se define la clase Persona que hereda de object
    def __init__(self, name, age):      #se hace su constructor que recibe dos parámetros name y age
        self.name = name                #valores de entrada
        self.age = age
    
    def birth(self):                    #el metodo birth no recibe parámetros de entrada
        self.age = self.age + 1
```

Crear una persona e imprimir sus atributos

```python
jugador = Persona("Juan", 21)                       #se crea un objeto de la clase Persona llamado jugador con nombre "Juan" y edad 21

print("el nombre del jugador es", jugador.name)     #se imprime el nombre del jugador con el atributo name del objeto jugador
print("la edad del jugador es", jugador.age)
```

Crea un jugador nuevo, y cambiale el nombre

```python
jugador2 = Persona("Camilo", 19)                        #se crea un otro objeto de la clase Persona

print("Nombre antes del cambio:", jugador2.name)

jugador2.name = "Murcia"                                #se cambia el nombre del jugador2 a "Murcia"
print("nombre despues del cambio:", jugador2.name)      #se imprime el nombre del jugador2 después del cambio
```

El primer jugador cumplió años, usa el metodo "Birth" para cambiar su edad.

```python
print("Edad antes del cumpleaños:", jugador.age)        #se imprime la edad del jugador

jugador.birth()                                         #metodo para cambiar la edad del jugador en 1
print("Edad despues del cumpleaños:", jugador.age)      #imprimimos la edad del jugador otra vez
```
