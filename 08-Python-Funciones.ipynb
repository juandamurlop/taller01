<h2>Introducción</h2>
En programacion, una funcion es un bloque de codigo diseñado para realizar una tarea específica, que puede ser reutilizado en distintas partes de un programa o incluso entre diferentes programas. Las funciones permiten organizar el código de manera más clara y estructurada, facilitando la resolución de problemas complejos al dividirlos en tareas más pequeñas y manejables.

Cree una función que divida la primera entrada por la segunda entrada:

```python
def divi(a, b):     #se crea la función divi que recibe dos parámetros a y b
    return(a/b)
```

Utilice la función <code> con </code> para la siguiente pregunta.

```python
def con(a, b):  #se crea la función con que recibe dos parámetros a y b
    return(a + b)
```

¿Se puede usar la función <code> con </code> que definimos antes para agregar números enteros o cadenas?

```python
#se puede concatenar strings y sumar números, dependiendo del tipo de dato que se le pase a la funcion

print(con(2, 5))
print(con("Hola ", "Juan"))
```

¿Se puede usar la función <code> con </code> que definimos antes para concentrar una lista o tupla?

```python
#tambien se puede usar para concatenar listas, ya que ambas entradas harán lo que pida la función, ejemplo

print(con(["Hola ", "Juan"], [18, "A"]))
```
