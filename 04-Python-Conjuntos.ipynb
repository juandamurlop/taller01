<h2>Introducción</h2>
En Python un conjunto es una estructura de datos que permite almacenar una colección de elementos de manera no ordenada y sin elementos duplicados, facilitando el manejo y la organización de información.


Transformar la lista <code>['rap','house','electronic music', 'rap']</code> en un conjunto:

```python
Conjunto = set(['rap','house','electronic music','rap'])
#se crea un conjunto llamado "Conjunto" que contiene los elementos 'rap', 'house', 'electronic music' y 'rap'. Los conjuntos no permiten elementos duplicados, por lo que el segundo 'rap' será ignorado.

Conjunto
```

Considera que la lista <code>A = [1, 2, 2, 1]</code> y el conjunto <code>B = set([1, 2, 2, 1])</code> suman <code>sum(A) = sum(B)</code>

```python
A = [1, 2, 2, 1]                            #creación lista "A"
B = {1, 2, 2, 1}                            #Creación conjunto "B"

print("la suma de la lista A es: ", sum(A))
print("la suma del conjunto B es: ", sum(B))

#aqui se ve algo interesante ya que la suma de A y de B dan resultados distintos , aunque parezca que sumamos lo mismo no es asi, ya que sirecordamos el conjunto "elimina" los duplicados asi que en verdad está sumando solo 2 + 1, ya que se encuentra mas de una vez en el conjunto
```

Crea un nuevo conjunto <code>album_set3</code> que sea la unión de <code>album_set1</code> y <code>album_set2</code>:

```python
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])                        #Creación conjunto 1
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])      #Creación conjunto 2

album_set3 = album_set1.union(album_set2)                                       #Creación conjunto 3, siendo una unión del 1 y 2
print("Al album 3 lo conforman: ", album_set3)                                  #impresión en pantalla del conjunto 3

#con ayuda del metodo "union" logramos unir abmos conjuntos y asignarlos al nuevo
```

Averigua si <code>album_set1</code> es un subconjunto de <code>album_set3</code>:

```python
album_set1.issubset(album_set3)     #verificamos directamente el subconjunto con "issubset"
```
